```python

from nltk.tokenize import word_tokenize

sentence = "The boy is playing football."

tokens = word_tokenize(sentence)

print(nltk.pos_tag(tokens))
```

```python
import string

tokens_without_punctuation = [w for w in tokens if w not in string.punctuation]
print(tokens_without_punctuation)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_words = [w for w in tokens_without_punctuation if w not in stop_words]
print(filtered_words)
```

```python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer

sentence = "The boy is playing football."

porter = PorterStemmer()
porter_result = [porter.stem(word) for word in sentence.split()]
porter_result
```

```python
lancaster = LancasterStemmer()
lancaster_result = [lancaster.stem(word) for word in sentence.split()]
lancaster_result
```

```python
snowball = SnowballStemmer('english')
snowball_result = [snowball.stem(word) for word in sentence.split()]
snowball_result
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_words = []

for word in filtered_words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print(lemmatized_words)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from nltk.corpus import wordnet as wn

syn = wn.synsets("car")[0]

G = nx.DiGraph()

G.add_node("car")

for h in syn.hypernyms():
    G.add_edge("car", h.name())

for h in syn.hyponyms()[:10]:
    G.add_edge("car", h.name())

nx.draw(G, with_labels=True, node_size=2500, font_size=9)

plt.title("WordNet Relationships of Car")
plt.show()
```
