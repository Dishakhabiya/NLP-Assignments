```python
def analyze_wordnet(word_freq, wordnet):
    results = []

    for word, frequency in word_freq.items():
        try:
            synsets = wordnet.synsets(word)
            synset_count = len(synsets)

            results.append({
                "word": word,
                "frequency": frequency,
                "synset_count": synset_count,
                "found": synset_count > 0
            })

        except Exception:
            results.append({
                "word": word,
                "frequency": frequency,
                "synset_count": 0,
                "found": False
            })

    return results
```

```python
def synset_distribution(results):
    one = 0
    two = 0
    three = 0
    four_plus = 0

    for item in results:
        count = item["synset_count"]

        if count == 1:
            one += 1
        elif count == 2:
            two += 1
        elif count == 3:
            three += 1
        elif count >= 4:
            four_plus += 1

    return [one, two, three, four_plus]
```

```python
from nltk.corpus import wordnet as wn
import networkx as nx
import matplotlib.pyplot as plt
import textwrap

# Create graph
G = nx.DiGraph()

selected = []

# Find 5 WordNet hypernym relations
for word, freq in french_freq.most_common(50):

    synsets = wn.synsets(word, lang="fra")

    for synset in synsets:

        hypernyms = synset.hypernyms()

        if hypernyms:

            hypernym = hypernyms[0]

            lemmas = hypernym.lemma_names(lang="fra")

            if lemmas:

                hypernym_name = lemmas[0].replace("_", " ")

                if (
                    word != hypernym_name
                    and (word, hypernym_name) not in selected
                ):
                    selected.append((word, hypernym_name))
                    break

    # Keep graph clean
    if len(selected) == 5:
        break


# Print selected relations
print("Selected WordNet Relations:\n")

for word, hypernym in selected:
    print(word, "→", hypernym)


# Add nodes and edges
for word, hypernym in selected:
    G.add_node(word)
    G.add_node(hypernym)
    G.add_edge(word, hypernym)


# Fixed two-column layout
pos = {}

for i, (word, hypernym) in enumerate(selected):
    pos[word] = (0, -i)
    pos[hypernym] = (2.8, -i)


# Wrap long labels so they fit better inside nodes
labels = {}

for node in G.nodes():
    labels[node] = "\n".join(
        textwrap.wrap(node, width=14)
    )


# Create figure
plt.figure(figsize=(13, 8))


# Draw all nodes - dark blue and bigger
nx.draw_networkx_nodes(
    G,
    pos,
    node_size=3500,
    node_color="darkblue"
)


# Draw arrows
nx.draw_networkx_edges(
    G,
    pos,
    arrows=True,
    arrowsize=18,
    width=1.5,
    edge_color="black"
)


# White text inside nodes
nx.draw_networkx_labels(
    G,
    pos,
    labels=labels,
    font_size=9,
    font_weight="bold",
    font_color="white"
)


# Column headings
plt.text(
    0,
    0.6,
    "French Corpus Words",
    ha="center",
    fontsize=13,
    fontweight="bold"
)

plt.text(
    2.8,
    0.6,
    "WordNet Hypernyms",
    ha="center",
    fontsize=13,
    fontweight="bold"
)


# Title
plt.title(
    "WordNet Relations from Scraped French Corpus",
    fontsize=17,
    fontweight="bold"
)

plt.axis("off")
plt.tight_layout()
plt.show()
```

### Hypernym/Hyponym Hierarchy for 'intelligence' (French)

This graph visualizes the hierarchical 'is-a' relationships for the French word 'intelligence' in WordNet.

*   **Hypernyms** are more general concepts.
*   **Hyponyms** are more specific concepts.

The graph shows how 'intelligence' is categorized, moving from broader concepts (up the hierarchy) to more specialized ones (down the hierarchy). Each node represents a synset, labeled with its primary lemma, and edges indicate the direction of generalization (hypernymy) or specialization (hyponymy).

```python
import networkx as nx
import matplotlib.pyplot as plt
from nltk.corpus import wordnet as wn

# Choose a Spanish word
search_word = 'inteligencia'
lang_code = 'spa'

# Get synsets for the word
synsets = wn.synsets(search_word, lang=lang_code)

if not synsets:
    print(f"No synsets found for '{search_word}' in {lang_code} WordNet.")
elif not any(s.pos() == 'n' for s in synsets):
    print(f"No noun synsets found for '{search_word}' in {lang_code} WordNet. Hypernym/hyponym relations are primarily for nouns.")
else:
    # Select a relevant noun synset (e.g., the first noun synset)
    target_synset = next(s for s in synsets if s.pos() == 'n')

    G = nx.DiGraph()

    def add_synset_to_graph(graph, synset, level=0, max_level=2, relationship_type=None):
        if level > max_level:
            return

        lemma_name = synset.lemmas(lang=lang_code)[0].name() if synset.lemmas(lang=lang_code) else synset.name()
        node_label = f"{lemma_name.capitalize()}.{synset.pos()}.{synset.lexname()[-2:]}"
        graph.add_node(synset, label=node_label, level=level)

        if relationship_type:
            graph.add_edge(relationship_type[0], synset, relation=relationship_type[1])

        # Add hypernyms (more general concepts)
        for hypernym in synset.hypernyms():
            add_synset_to_graph(graph, hypernym, level + 1, max_level, (synset, 'is-a (hypernym)'))

        # Add hyponyms (more specific concepts)
        for hyponym in synset.hyponyms():
            add_synset_to_graph(graph, hyponym, level + 1, max_level, (synset, 'has-type (hyponym)'))


    # Build the graph starting from the target synset
    add_synset_to_graph(G, target_synset, level=0, max_level=2) # Explore 2 levels up and down

    plt.figure(figsize=(18, 12))

    # Use a hierarchical layout
    pos = nx.multipartite_layout(G, subset_key='level', align='vertical')

    # Fallback to spring layout if multipartite fails or looks bad
    if not pos:
         pos = nx.spring_layout(G, k=0.9, iterations=50, seed=42)

    node_labels = nx.get_node_attributes(G, 'label')
    node_colors = ['#FFD700' if node == target_synset else '#ADD8E6' for node in G.nodes()]

    nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=3500, alpha=0.9)
    nx.draw_networkx_labels(G, pos, labels=node_labels, font_size=8, font_weight='bold')

    # Draw edges and label them
    edge_labels = {}
    for u, v, data in G.edges(data=True):
        edge_labels[(u, v)] = data.get('relation', '')

    nx.draw_networkx_edges(G, pos, edge_color='gray', width=1, alpha=0.6, arrows=True, arrowsize=15)
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='red', font_size=7)

    plt.title(f"Hypernym/Hyponym Hierarchy for '{search_word.capitalize()}' (Spanish) Synset: {target_synset.name()}", fontsize=16)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```
